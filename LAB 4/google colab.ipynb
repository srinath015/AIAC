Task Description#1

● Zero-shot: Prompt AI to write a function that checks whether a given year is a leap year.

Expected Output#1

● AI-generated function with no examples provided

```
def is_leap(year):
  """
  Checks if a given year is a leap year.

  Args:
    year: An integer representing the year.

  Returns:
    True if the year is a leap year, False otherwise.
  """
  if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    return True
  else:
    return False

# Read year from console
year_str = input("Enter a year: ")

try:
  year = int(year_str)
  if is_leap(year):
    print(f"{year} is a leap year.")
  else:
    print(f"{year} is not a leap year.")
except ValueError:
  print("Invalid input. Please enter a valid year as a number.")
```

Task Description#2

●	One-shot: Give one input-output example to guide AI in writing a function that converts centimeters to inches.

Expected Output#2

●	Function with correct conversion logic


```
# Conversion factor: 1 cm = 0.39 inches
CM_TO_INCH = 0.39

# Read centimeters from console
cm_str = input("Enter length in centimeters: ")

try:
  cm = float(cm_str)
  inches = cm * CM_TO_INCH
  print(f"{cm} centimeters is equal to {inches:.2f} inches.")
except ValueError:
  print("Invalid input. Please enter a valid number for centimeters.")
```

Task Description#3

●	Few-shot: Provide 2–3 examples to generate a function that formats full names as “Last, First”.

Expected Output#3

   .     Well-structured function respecting the examples



```
# Read the full name from the console
full_name = input("Enter full name: ")

# Split the full name into parts (assuming space separation)
name_parts = full_name.split()

# Check if there are at least two parts (first and last name)
if len(name_parts) >= 2:
  # Assume the last part is the last name and the first part is the first name
  last_name = name_parts[-1]
  first_name = name_parts[0]

  # Format as "Last, First"
  formatted_name = f"{last_name}, {first_name}"
  print(f"Formatted name: {formatted_name}")
else:
  print("Invalid input. Please enter a full name with at least a first and last name.")
```

```
# Read the string from the console
input_string = input("Enter a string: ")

# Define a set of vowels
vowels = "aeiouAEIOU"

# Initialize vowel count
vowel_count = 0

# Iterate through the string and count vowels
for char in input_string:
  if char in vowels:
    vowel_count += 1

# Print the result
print(f"The number of vowels in the string is: {vowel_count}")
```

Task Description#5

●	Use few-shot prompting to generate a function that reads a .txt file and returns the number of lines.

Expected Output#5

●	Working file-processing function with AI-guided logic


```
def count_lines_in_file(filepath):
  """
  Reads a text file and returns the number of lines.

  Args:
    filepath: The path to the text file.

  Returns:
    The number of lines in the file, or -1 if the file cannot be opened.
  """
  try:
    with open(filepath, 'r') as file:
      line_count = sum(1 for line in file)
    return line_count
  except FileNotFoundError:
    print(f"Error: The file '{filepath}' was not found.")
    return -1
  except Exception as e:
    print(f"An error occurred: {e}")
    return -1

# Specify the file path
file_path = r"/SR UNIVERSITY.txt" # Using raw string to handle backslashes

# Count the lines and display the result
number_of_lines = count_lines_in_file(file_path)

if number_of_lines != -1:
  print(f"The file '{file_path}' has {number_of_lines} lines.")
```
